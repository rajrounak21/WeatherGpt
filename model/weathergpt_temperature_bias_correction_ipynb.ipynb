```python
from datasets import load_dataset

ds = load_dataset("Arko007/weathergpt-d1-mos-dataset")
print(ds)
print(ds["train"].num_rows)
print(ds["train"].column_names)
```

```python
print(ds["train"].num_rows)
print(ds["train"].column_names)
```

```python
train = ds["train"]
```

```python
temperature_data = train.select_columns([
    "fc_temperature_2m_gfs_seamless",
    "fc_relative_humidity_2m_gfs_seamless",
    "fc_wind_speed_10m_gfs_seamless",
    "lead_hours",
    "lat",
    "lon",
    "elevation_m",
    "hour_utc",
    "month",
    "valid_time",
    "truth_temperature_2m"
])

print(temperature_data)
```

```python
print(temperature_data.column_names)
print(temperature_data[:5])
```

```python
print("Rows:", temperature_data.num_rows)
print("Columns:", temperature_data.num_columns)
```

```python
for column in temperature_data.column_names:
    missing = sum(x is None for x in temperature_data[column])
    print(column, "→", missing)
```

```python
for i in range(5):
    gfs = temperature_data[i]["fc_temperature_2m_gfs_seamless"]
    truth = temperature_data[i]["truth_temperature_2m"]
    error = truth - gfs

    print(
        f"GFS: {gfs}°C | "
        f"Truth: {truth}°C | "
        f"Error: {error:.2f}°C"
    )
```

```python
def add_temeprature_error(batch):
  batch["temperature_error"] = [
      truth - gfs
      for truth, gfs in zip(
          batch["truth_temperature_2m"],
          batch["fc_temperature_2m_gfs_seamless"]
          )
  ]
  return batch

temperature_data = temperature_data.map(
    add_temeprature_error,
    batched = True
)
```

```python
print(len(temperature_data.column_names))
print(temperature_data[:5])
```

```python
print("First:", min(temperature_data["valid_time"]))
print("Last:", max(temperature_data["valid_time"]))
```

```python
times = sorted(set(temperature_data["valid_time"]))

train_end = times[int(len(times) * 0.70)]
val_end = times[int(len(times) * 0.85)]

print("Train ends:", train_end)
print("Validation ends:", val_end)
```

```python
train_data = temperature_data.filter(
    lambda row: row["valid_time"] < train_end
)

val_data = temperature_data.filter(
    lambda row: train_end <= row["valid_time"] < val_end
)

test_data = temperature_data.filter(
    lambda row: row["valid_time"] >= val_end
)
```

```python
print("Train:", train_data.num_rows)
print("Validation:", val_data.num_rows)
print("Test:", test_data.num_rows)
```

```python
print("Train:", min(train_data["valid_time"]), "→", max(train_data["valid_time"]))
print("Validation:", min(val_data["valid_time"]), "→", max(val_data["valid_time"]))
print("Test:", min(test_data["valid_time"]), "→", max(test_data["valid_time"]))
```

```python
import numpy as np

gfs_test = np.array(test_data["fc_temperature_2m_gfs_seamless"])
truth_test = np.array(test_data["truth_temperature_2m"])

mae_gfs = np.mean(np.abs(truth_test - gfs_test))

print("GFS MAE:", mae_gfs)
```

```python
rmse_gfs = np.sqrt(
    np.mean((truth_test - gfs_test) ** 2)
)

print("GFS RMSE:", rmse_gfs)
```

```python
features = [
    "fc_temperature_2m_gfs_seamless",
    "fc_relative_humidity_2m_gfs_seamless",
    "fc_wind_speed_10m_gfs_seamless",
    "lead_hours",
    "lat",
    "lon",
    "elevation_m",
    "hour_utc",
    "month"
]

target = "temperature_error"
```

```python
X_train = np.column_stack([
    np.array(train_data[col], dtype=np.float32)
    for col in features
])
```

```python
X_val = np.column_stack([
    np.array(val_data[col], dtype=np.float32)
    for col in features
])
```

```python
X_test = np.column_stack([
    np.array(test_data[col], dtype=np.float32)
    for col in features
])
```

```python
y_train = np.array(train_data[target], dtype=np.float32)
y_val = np.array(val_data[target], dtype=np.float32)
y_test = np.array(test_data[target], dtype=np.float32)


```

```python
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
from xgboost import XGBRegressor

model = XGBRegressor(
    n_estimators=500,
    max_depth=8,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    tree_method="hist",
    n_jobs=-1
)

model.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    verbose=50
)
```

```python
predicted_error = model.predict(X_test)

gfs_test = np.array(test_data["fc_temperature_2m_gfs_seamless"])

corrected_temperature = gfs_test + predicted_error

truth_test = np.array(test_data["truth_temperature_2m"])

mae_corrected = np.mean(
    np.abs(truth_test - corrected_temperature)
)

rmse_corrected = np.sqrt(
    np.mean((truth_test - corrected_temperature) ** 2)
)

print("Corrected GFS MAE:", mae_corrected)
print("Corrected GFS RMSE:", rmse_corrected)
```

```python
import pickle

with open("weathergpt_temperature_xgb.pkl", "wb") as f:
    pickle.dump(model, f)

print("Model saved successfully!")
```
